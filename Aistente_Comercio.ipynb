```python
import google.generativeai as genai
import os
import time
import json
import getpass

api_key = os.environ.get('GEMINI_API_KEY')
if not api_key:
    api_key = getpass.getpass("Ingresá tu GEMINI_API_KEY: ")

genai.configure(api_key=api_key)
print("Cliente Gemini configurado correctamente.")
```

```python
def cargar_prompt(ruta):
    with open(ruta, encoding='utf-8') as f:
        return f.read()

def cargar_stock(ruta='products.json'):
    with open(ruta, encoding='utf-8') as f:
        return json.load(f)

SYSTEM_PROMPT_V1 = cargar_prompt('system_promptV1.txt')
SYSTEM_PROMPT_V2 = cargar_prompt('system_promptV2.txt')
STOCK = cargar_stock()

print("Archivos cargados correctamente.")
```

```python
def probar_mostrador(prompt_version, consulta):
    modelo = genai.GenerativeModel(model_name="gemini-3.8-flash", system_instruction=prompt_version)
    
    #stock disponible
    consulta_completa = f"{consulta}\n\n[STOCK DISPONIBLE]\n{json.dumps(STOCK, indent=2)}"
    
    print(f"--- CLIENTE DICE ---\n'{consulta}'\n")
    inicio = time.time()
    respuesta = modelo.generate_content(consulta_completa)
    tiempo = round(time.time() - inicio, 2)
    
    print("--- RESPUESTA DEL ASISTENTE ---")
    print(respuesta.text)
    print(f"\n[Métricas: {tiempo} segundos | {respuesta.usage_metadata.total_token_count} tokens]")
    print("-" * 50)
```

```python
# PRUEBA 1: Usando la Versión 1 del Prompt
probar_mostrador(SYSTEM_PROMPT_V1, "Necesito instalar un aire de 3500 frigorías. ¿Qué llevo?")
```

```python
# PRUEBA 2: Usando la Versión 2 del Prompt
probar_mostrador(SYSTEM_PROMPT_V2, "Hola, dame cable y térmica para enchufar unas cosas en el taller.")
```
